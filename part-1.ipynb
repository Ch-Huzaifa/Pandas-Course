### Series in Pandas:
A Series is a one-dimensional labeled array that can store a single column of data with a unique label called Index starting from 0.  
In the example:  
Values = 10, 20, 30, 40  
Index labels = 0, 1, 2, 3

```python
s = pd.Series([10, 20, 30, 40])
print(s)
```

### DataFrame in Pandas

A DataFrame is a two-dimensional table with rows and columns, similar to an Excel spreadsheet or database table.
In the example :

 ## Describe()
 The describe method is used to get a quick statistical summary of a Series or DataFrame.

| --------- | ------------------------------------------------ |
| **count** | Number of non-missing values in each column.                     |
| **mean**  | Average of all the values in a column.                                   |
| **std**   | Standard deviation is low if values are closed and viceversa.             |
| **min**   | Smallest value from the column.                                  |
| **25%**   | It give the value that saperates the First quartile.(25% of data is below this value) |
| **50%**   | It gives the value that saperate the data set into half (Midddle value).            |
| **75%**   | It gives the value that saperate the Third quartile (75% of data is below this value) |
| **max**   | Largest value from the column.                                   |

<br>
The **columns** attribute returns the names of all columns in a DataFrame.

### Selecting Single and Multiple Comments. 

```python
data={
    "Name":['Huzaifa','Shuban','Taimoor','Ahsan','Ali'],
    "Age":[20,21,21,22,35],
    "Arid_no":['24-Arid-801','24-Arid-848','24-Arid-852','24-Arid-791','24-Arid-795'],
    "City":['Rawalpindi','Mansehra','KotliSattiyan','Rawalpindi','Chakwal'],
    "Salary":[80000,75000,70000,20000,5000]
}

df=pd.DataFrame(data)
print("Selecting a single column:\n")
print(df["Name"])
print("\n<<<<<<<<<<<<<<<<<<>>>>>>>>>>>>>>>>>>>>>")
print("\nSelecting multiple columns:\n")
print(df[["Name","Salary"]])
```

 ---
### Filtering Rows with Single and Multiple Conditions.

```python
data={
    "Name":['Huzaifa','Shuban','Taimoor','Ahsan','Ali'],
    "Age":[20,21,21,22,35],
    "Arid_no":['24-Arid-801','24-Arid-848','24-Arid-852','24-Arid-791','24-Arid-795'],
    "City":['Rawalpindi','Mansehra','KotliSattiyan','Rawalpindi','Chakwal'],
    "Salary":[80000,75000,70000,20000,5000]
}

df=pd.DataFrame(data)
condition1=df[df["Salary"] < 50000] # This will return the rows.
print("Row Filtering with Single Condition.\n") 
print(condition1,"\n")

condition2=df[(df["Salary"] > 50000)&(df["Age"]<= 20)]
print("Row Filtering with Multiple Conditions.\n") 
print(condition2,"\n")
```
