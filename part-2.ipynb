## Detecting Missing Values:
### 1.isnull() in Pandas:
The **isnull()** method checks whether values are missing (NaN or None)and  creates a table of True and False  
It returns:  
True for missing values and False for non-missing values.

### 2.isnull().sum() in Pandas:
The sum() method counts the True values and returns a series for each column.

### 3.isnull().sum().sum() in Pandas:
.sum().sum() Counts total missing values in the entire DataFrame.

```python
data={
    "Name":['Huzaifa','Shuban','Taimoor','Ahsan',None],
    "Age":[20,21,21,22,None],
    "Arid_no":['24-Arid-801','24-Arid-848','24-Arid-852','24-Arid-791',None],
    "City":['Rawalpindi','Mansehra','KotliSattiyan','Rawalpindi',None],
    "Salary":[80000,75000,70000,20000,None]
}

df=pd.DataFrame(data)
print("Working of isnull()\n")
print(df.isnull())
print("\nWorking of isnull().sum()\n")
print(df.isnull().sum())
print("\nWorking of isnull().sum().sum()\n")
print(df.isnull().sum().sum())
```

___


## Removing Missing Values:
### dropna() in Pandas:
The dropna() method is used to remove rows or columns that contain missing values (NaN).It removes the complete row containing missing values as axis=0 by default.  
**df.dropna(axis=1)** will only remove the column contining missing values.  

```python
data={
    "Name":['Huzaifa','Shuban','Taimoor','Ahsan','Ali'],
    "Age":[20,21,None,22,35],
    "Arid_no":['24-Arid-801','24-Arid-848','24-Arid-852','24-Arid-791','24-Arid-795'],
    "City":['Rawalpindi','Mansehra','KotliSattiyan','Rawalpindi','Chakwal'],
    "Salary":[80000,75000,70000,20000,None]
}

df=pd.DataFrame(data)

print("The row 3 and 5 will be removed because they conatin missing values\n")
print(df.dropna())

print("\nThe column Age and Salary will be removed because they conatin missing values\n")
print(df.dropna(axis=1))
```

___
### Using dropna(subset=[...])
It  will Checks only the specified column(s) for missing values.  

**Example** :  
If df.dropna(subset=['Age'])  

| Name  | Age | City    |
| ----- | --- | ------- |
| Ali   | 20  | Lahore  |
| Sara  | NaN | Karachi |
| Ahmed | 22  | NaN     |


Result:

| Name  | Age | City   |
| ----- | --- | ------ |
| Ali   | 20  | Lahore |
| Ahmed | 22  | NaN    |


Only Sara is removed because Age is missing. Ahmed stays because Age is present, even though City is missing.

```python
data={
    "Name":['Huzaifa','Shuban','Taimoor','Ahsan','Ali'],
    "Age":[20,21,None,22,35],
    "Arid_no":['24-Arid-801','24-Arid-848','24-Arid-852','24-Arid-791','24-Arid-795'],
    "City":['Rawalpindi','Mansehra','KotliSattiyan','Rawalpindi','Chakwal'],
    "Salary":[80000,75000,70000,20000,None]
}

df=pd.DataFrame(data)

print("This will remove the rows in which Age is missing.\n")
print(df.dropna(subset=['Age']))
```

## Filling Missing values:
  
  | Method                      | Purpose                      |
| --------------------------- | ---------------------------- |
| df.fillna(0)              | Replace NaN with 0           |
| df.fillna("Unknown")      | Replace NaN with text        |
| df.fillna(df.mean())      | Replace NaN with column mean |
| df.ffill()                | Fill with previous value     |
| df.bfill()                | Fill with next value         |


```python
data = {
    "Name": ['Huzaifa', 'Shuban', None, 'Ahsan', None],
    "Age": [20, None, 21, 22, None],
    "Arid_no": ['24-Arid-801', None, '24-Arid-852', '24-Arid-791', None],
    "City": ['Rawalpindi', 'Mansehra', None, 'Rawalpindi', None],
    "Salary": [80000, 75000, None, 20000, None]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("\n" + "="*50 + "\n")

print("fillna('Unknown'):")
print(df.fillna("Unknown"))
print("\n" + "="*50 + "\n")

print("fillna() with dictionary:")
print(df.fillna({
    "Name": "No Name",
    "Age": 0,
    "Arid_no": "Not Assigned",
    "City": "Unknown City",
    "Salary": 0
}))
print("\n" + "="*50 + "\n")

df["Age"] = df["Age"].fillna(df["Age"].mean())
print("Fill Age NaN with Mean:")
print(df)
print("\n" + "="*50 + "\n")

print("Forward Fill (ffill):")
print(df.ffill())
print("\n" + "="*50 + "\n")

print("Backward Fill (bfill):")
print(df.bfill())
print("\n" + "="*50 + "\n")

df["City"] = df["City"].fillna("Islamabad")
print("Fill City column with Islamabad:")
print(df)
print("\n" + "="*50 + "\n")
```

# Interpolation in Pandas: 
 Interpolation is a technique used to fill missing values by estimating them from the surrounding data points.Instead of replacing missing values with a fixed number (like 0 or the mean), interpolation calculates a value that fits the trend of the data.  
 Example: [70, NaN, 90]  
 Output : [70, 80, 90].....It will fill the middle value with 80.

## Advantages of Using Interpolation:
1. Maintains Data Trends
2. More Realistic Values
3. Prevents Data Loss
4. Works well with Numeric data.
5. Improves Analysis Accuracy
6. Preserves Relationships Between Values

## Limitation:
Interpolation works best when the missing values are between known values and there is a meaningful pattern. If many values are missing or the data is highly irregular, other methods may be more appropriate.

```python
data = {
    "id": [1, None, 3, None, 5],
    "Name": ['Huzaifa', 'Shuban', 'Taimoor', 'Ahsan', 'Ali'],
}

df = pd.DataFrame(data)
df['id']=df['id'].interpolate()
print(df)
```

### The sort_values() method is used to sort a DataFrame by one or more columns.   

## Sorting A Single Column:
 
**Syntax** : df.sort_values(by="Column_Name", ascending=True/False , inplace=True)  
(ascending=true) will sort in ascending order and (ascending =False) will sort in descending order.  

## Sort by Multiple Columns:
**Syntax** : df.sort_values(by=["Age", "Salary"],ascending=[True,False])  
Pandas first sorts by Age. If two rows have the same age, it then sorts by Salary. 

```python
data ={
    "Name": [ "Shuban","Huzaifa", "Taimoor", "Ali", "Ahsan"],
    "Age": [21, 20, 21, 35, 22 ],
    "Salary": [ 75000, 80000, 78000, 20000, 50000 ]
}

df=pd.DataFrame(data)

print("Original Dataframe\n")
print(df)
print("\n" + "="*50 + "\n")

print ("Sorting by Age:\n")
print(df.sort_values(by="Age", ascending=True))
print("\n" + "="*50 + "\n")

print("\nSorting using Age and Salary\n")
print(df.sort_values(by=["Age","Salary"], ascending=[True,False]))
print("\n" + "="*50 + "\n")
```
