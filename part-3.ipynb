## Data Aggregation!!!
___

### Aggregation means summarizing multiple data values into a single value.

| Method      | Purpose                   |
| ----------- | ------------------------- |
| sum()     | Total                     |
| mean()    | Average                   |
| min()     | Smallest value            |
| max()     | Largest value             |
| median()  | Middle value              |
| agg()     | Multiple aggregations     |

```python
data={
    "Name": ["Huzaifa", "Shuban", "Taimoor", "Ahsan"],
    "Salary": [80000, 75000, 70000, 20000]
}

df=pd.DataFrame(data)

print("\nSum of Salary:")
print(df["Salary"].sum())

print("\nMean Salary:")
print(df["Salary"].mean())

print("\nMinimum Salary:")
print(df["Salary"].min())

print("\nMaximum Salary:")
print(df["Salary"].max())

print("\nMedian Salary:")
print(df["Salary"].median())

print("\nMultiple Aggregations using agg():")
print(df["Salary"].agg(["sum", "mean", "min", "max", "count", "median"]))
```

___
### We can Aggregate a column with respect to another column using groupby() method.
This example will show how **groupby()** calculates the sum, mean, min, max, and count separately for the IT and HR departments.  
  
Groups Created:

IT Department

Huzaifa → 80000, Shuban → 75000, Ali → 60000, Warda → 65000

HR Department

Taimoor → 70000, Ahsan → 20000, Tuba → 55000, Hadia → 45000

```python
df = pd.DataFrame({
    "Name": ["Huzaifa", "Shuban", "Taimoor", "Ahsan", "Ali", "Tuba", "Warda", "Hadia"],
    "Department": ["IT", "IT", "HR", "HR", "IT", "HR", "IT", "HR"],
    "Salary": [80000, 75000, 70000, 20000, 60000, 55000, 65000, 45000]
})

print("Original DataFrame:\n")
print(df)

print("\n" + "="*50)

print("\nMultiple Aggregations by Department:\n")
print(df.groupby("Department")["Salary"].agg(["sum", "mean", "min", "max", "count"]))
print("\n" + "="*50)
```

# Step 8 :
## DataSets Merging and Joining!!!
___

### **pd.merge()** is used to combine two DataFrames based on one or more common columns, similar to SQL JOIN operations.

**Syntax** : pd.merge(df1, df2, on="common_column")  

```python
students = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Huzaifa", "Shuban", "Taimoor", "Ahsan"]
})

print("This is the Student Dataframe\n\n",students)
print("\n" + "="*50)

marks = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Marks": [85, 90, 78, 88]
})

print("\nThis is the Marks Dataframe\n\n",marks)
print("\n" + "="*50)

result = pd.merge(students, marks, on="ID")
print("\nThis is the Combined Dataframe\n\n",result)
print("\n" + "="*50)
```

___
### We also use different merge types to control which rows are kept and displayed when combining two DataFrames.

| Merge Type | Keeps                                         |
| ---------- | --------------------------------------------- |
| **Inner**  | Only matching rows from both DataFrames       |
| **Left**   | All rows from left + matching rows from right |
| **Right**  | All rows from right + matching rows from left |
| **Outer**  | All rows from both DataFrames                 |
| **Cross**  | Every possible combination of rows            |

**Syntax** : pd.merge(left_dataframe, right_dataframe, on="common_column", how="merge_type")

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Huzaifa", "Shuban", "Taimoor", "Ahsan", "Ali"]
})

df2 = pd.DataFrame({
    "ID": [3, 4, 5, 6, 7],
    "Marks": [80, 75, 90, 88, 70]
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

print("\n" + "="*50)

print("\nInner Merge:")
print(pd.merge(df1, df2, on="ID", how="inner"))
print("\n" + "="*50)

print("\nLeft Merge:")
print(pd.merge(df1, df2, on="ID", how="left"))
print("\n" + "="*50)

print("\nRight Merge:")
print(pd.merge(df1, df2, on="ID", how="right"))
print("\n" + "="*50)

print("\nOuter Merge:")
print(pd.merge(df1, df2, on="ID", how="outer"))
print("\n" + "="*50)
```

# Step 9 :
## Concatenating The Datasets !!!
___

### pd.concat() is used to combine multiple DataFrames either by stacking rows or joining columns without matching keys.
**Synatx** : pd.concat([df1, df2], axis=0, ignore_index=True)  

| Axis   | Meaning            | Result       |
| ------ | ------------------ | ------------ |
| axis=0 | Vertical stacking  | Adds rows    |
| axis=1 | Horizontal joining | Adds columns |

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Huzaifa", "Shuban", "Taimoor", "Ahsan"]
})

df2 = pd.DataFrame({
    "ID": [6, 7, 8],
    "Name": ["Tuba", "Warda", "Hadia"]
})

print("DataFrame 1:\n")
print(df1)

print("\nDataFrame 2:\n")
print(df2)
print("\n" + "="*50)

row_concat = pd.concat([df1, df2], axis=0, ignore_index=True)
print("\nRow-wise Concatenation (axis=0):")
print(row_concat)
print("\n" + "="*50)

col_concat = pd.concat([df1, df2], axis=1)
print("\nColumn-wise Concatenation (axis=1):")
print(col_concat)
print("\n" + "="*50)
```
